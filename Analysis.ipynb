#student Internal reuslt analysis using python,pandas,matplotlib lib..

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#importing the csv file of student performance
df=pd.read_csv("students_result.csv")
```

```python
#calculating the percenatge of stduents
df["Total"]=df[["SEO","ADBMS","WWD","SAM","IKS","CN"]].sum(axis=1)
df["Percentage"]=round((df["Total"]/150)*100,2)
```

```python
#result 
df["Result"]=df["Percentage"].apply(lambda x:"Pass" if x>=40 else "Fail")
```
